## **Bước 1: Chuẩn hóa khoảng chặn thực tế $[L_i, U_i]$ và Lọc mâu thuẫn dữ liệu**
Thay vì gán cận dưới bằng 0 một cách máy móc (vốn không phản ánh đúng thực tế vì không có người đi làm hưởng lương nào có thu nhập bằng 0), ta thiết lập mức sàn tối thiểu kinh tế và kiểm tra tính hợp lệ.
**Quy ước khoảng chặn cho từng khách hàng $i$:**
- **Khoảng hữu hạn (VD: "Từ 20 - 50 triệu"):** $L_i = 20.000.000$, $U_i = 50.000.000$.
- **Khoảng mở dưới (VD: "Dưới 20 triệu"):** Cận dưới $L_i$ được đặt bằng mức sàn tối thiểu thực tế (ví dụ: sàn thu nhập tối thiểu là 5.000.000 VNĐ để tránh kéo dôi phân phối về 0), cận trên $U_i = 20.000.000$.
- **Khoảng mở trên (VD: "Từ 100 triệu VNĐ trở lên"):** Cận dưới $L_i = 100.000.000$, cận trên tạm thời đặt là $U_i = \infty$ để xử lý bằng đuôi Pareto.

**Bộ lọc mâu thuẫn tài chính (Logical Validation):**
Loại bỏ các bản ghi vi phạm định luật bảo toàn dòng tiền (ví dụ: tổng tỷ lệ chi tiêu cộng với tỷ lệ trả nợ vượt quá $100\%$ thu nhập nhưng lại khai báo khoản tích lũy dương lớn bất thường).

```python
import numpy as np
import pandas as pd
from scipy.stats import gaussian_kde

# Define columns for estimation based on the prompt
columns_to_estimate = [
    'Thu nhập',
    'Tỷ lệ chi tiêu',
    'Tích luỹ',
    'Tổng dư nợ',
    'Tổng nợ/TS',
    'Tài sản nhàn rỗi',
    'Tỷ lệ thanh toán Nợ',
    'Phí sẵn sàng trả',
    'Số tiền ban đầu'
]

# --- Part 1: Handle specific categorical values ---

# Values to be replaced with 0.00
zero_values = [
    "Hiện tại không có khoản nợ đáng kể",
    "Chưa có khoản dư đều đặn",
    "Chưa có khoản tiền nhàn rỗi",
    "Tôi không sẵn sàng trả phí" # Added based on df.head() for 'Phí sẵn sàng trả'
]

# Apply replacements to the specified columns
# It's better to do this *before* parsing the ranges, so the parser doesn't deal with these strings
for col in columns_to_estimate:
    if col in df.columns:
        # User requested to keep 'Chưa theo dõi cụ thể', 'Chưa tiện chia sẻ', 'Không tiện chia sẻ', 'Không cố định' as they are.
        # So, the blank_values replacement to np.nan is removed.
        df[col] = df[col].replace(zero_values, 0.00)

# Handle NaN values for 'Tổng nợ/TS' specifically after other replacements
if 'Tổng nợ/TS' in df.columns:
    df['Tổng nợ/TS'] = df['Tổng nợ/TS'].fillna(0.00)

# --- Part 2: Standardize range values for 'Thu nhập' and similar columns ---

# Function to parse range strings into (L_i, U_i) with unit handling
def parse_range_string(range_str_val, min_floor=None, is_percentage=False):
    if pd.isna(range_str_val):
        return np.nan, np.nan
    if range_str_val == 0.00: # Explicitly handle already zeroed values as exact 0.00
        return 0.00, 0.00

    range_str = str(range_str_val).strip()
    range_str_lower = range_str.lower()

    current_multiplier = 1.0
    if not is_percentage: # Only apply multipliers for non-percentage values
        if 'tỷ' in range_str_lower:
            current_multiplier = 1_000_000_000
            range_str_lower = range_str_lower.replace(' tỷ', '').replace('tỷ', '')
        elif 'triệu' in range_str_lower:
            current_multiplier = 1_000_000
            range_str_lower = range_str_lower.replace(' triệu', '').replace('triệu', '')
        range_str_lower = range_str_lower.replace(' vnđ', '').replace('.', '') # Remove VND and thousand separators

    # Helper for converting numerical string parts to float
    def safe_float_convert(s):
        try:
            return float(s)
        except ValueError:
            return np.nan

    if is_percentage:
        # Percentage specific parsing
        if 'dưới' in range_str_lower or 'trở xuống' in range_str_lower:
            val_str = range_str_lower.replace('dưới', '').replace('trở xuống', '').replace('/ năm', '').replace('%', '').strip()
            upper = safe_float_convert(val_str) / 100
            lower = min_floor if min_floor is not None else 0.0
            return lower, upper
        elif 'trên' in range_str_lower or 'trở lên' in range_str_lower:
            val_str = range_str_lower.replace('trên', '').replace('trở lên', '').replace('/ năm', '').replace('%', '').strip()
            lower = safe_float_convert(val_str) / 100
            upper = 1.0 # Max percentage is 100%
            return lower, upper
        elif '–' in range_str or '-' in range_str: # Handles 40-60%
            parts = range_str.replace('%', '').replace('–', '-').split('-')
            if len(parts) == 2:
                lower = safe_float_convert(parts[0].strip()) / 100
                upper = safe_float_convert(parts[1].strip()) / 100
                return lower, upper
        else: # Handle exact percentages if any, or single value like "50%"
            val = safe_float_convert(range_str.replace('%', '').strip())
            if not pd.isna(val):
                return val / 100, val / 100
    else: # Numerical values
        # Handle 'dưới X'
        if 'dưới' in range_str_lower:
            val_str = range_str_lower.replace('dưới', '').strip()
            upper = safe_float_convert(val_str) * current_multiplier
            lower = min_floor if min_floor is not None else 0.0
            return lower, upper
        # Handle 'trên X' or 'trở lên X'
        elif 'trở lên' in range_str_lower or 'trên' in range_str_lower:
            val_str = range_str_lower.replace('trở lên', '').replace('trên', '').replace('từ', '', 1).strip() # Remove 'từ' only if it's at the start
            lower = safe_float_convert(val_str) * current_multiplier
            upper = np.inf
            return lower, upper
        # Handle 'từ X - Y' or 'X - Y'
        elif ('từ' in range_str_lower and ('-' in range_str_lower or '–' in range_str_lower)) or ('-' in range_str_lower or '–' in range_str_lower):
            temp_str = range_str_lower.replace('từ', '', 1).strip().replace('–', '-') # Remove 'từ' only if it's at the start
            parts = temp_str.split('-')
            if len(parts) == 2:
                lower = safe_float_convert(parts[0].strip()) * current_multiplier
                upper = safe_float_convert(parts[1].strip()) * current_multiplier
                return lower, upper
            elif len(parts) == 1: # Fallback for 'từ X' if it's not 'từ X - Y' but caught here
                val = safe_float_convert(parts[0].strip()) * current_multiplier
                if not pd.isna(val):
                    return val, val
        # Handle single number values like "100" (after unit removal)
        else:
            val = safe_float_convert(range_str_lower) * current_multiplier
            if not pd.isna(val):
                return val, val
    return np.nan, np.nan # Default for unparseable strings

# Specific minimum floor for 'Thu nhập'
MIN_INCOME_FLOOR = 5_000_000

# Dictionary to configure parsing for each relevant column
# min_floor is used for "Dưới X" cases, is_percentage for handling percentage parsing
parsing_configs = {
    'Thu nhập': {'min_floor': MIN_INCOME_FLOOR, 'is_percentage': False},
    'Tỷ lệ chi tiêu': {'min_floor': 0.0, 'is_percentage': True},
    'Tích luỹ': {'min_floor': 0, 'is_percentage': False},
    'Tổng dư nợ': {'min_floor': 0, 'is_percentage': False},
    'Tổng nợ/TS': {'min_floor': 0.0, 'is_percentage': True},
    'Tài sản nhàn rỗi': {'min_floor': 0, 'is_percentage': False},
    'Tỷ lệ thanh toán Nợ': {'min_floor': 0.0, 'is_percentage': True},
    'Phí sẵn sàng trả': {'min_floor': 0.0, 'is_percentage': True},
    'Số tiền ban đầu': {'min_floor': 0, 'is_percentage': False},
}

# Apply parsing for all configured columns
for col, params in parsing_configs.items():
    if col in df.columns:
        df[[f'{col} - Lower', f'{col} - Upper']] = df[col].apply(
            lambda x: pd.Series(parse_range_string(x, min_floor=params['min_floor'], is_percentage=params['is_percentage']))
        )

# --- Step 2: Building KDE models ---

kde_models = {}

print("\n--- Step 2: Building KDE models ---")

for col in columns_to_estimate:
    lower_col = f'{col} - Lower'
    upper_col = f'{col} - Upper'

    if lower_col in df.columns and upper_col in df.columns:
        # We need data points for KDE. For bounded intervals, use midpoints.
        # Exclude infinite upper bounds (handled by Pareto in next step) and NaNs.

        # Identify rows where both bounds are finite and valid
        finite_bounded_mask = df[upper_col].notna() & (df[upper_col] != np.inf) & df[lower_col].notna()

        # Create a single representative point for each valid observation for KDE
        # Use lower bound if lower==upper (exact value), else use midpoint
        data_points = df[finite_bounded_mask].apply(
            lambda row: row[lower_col] if row[lower_col] == row[upper_col] else (row[lower_col] + row[upper_col]) / 2,
            axis=1
        )

        # Filter out any NaN values that might have resulted from parsing or empty ranges
        data_points = data_points.dropna()

        if len(data_points) > 1: # gaussian_kde requires at least 2 data points
            try:
                # Reshape for gaussian_kde (needs 2D array: (1, N) for 1D data)
                data_points_reshaped = data_points.values.reshape(1, -1)
                kde = gaussian_kde(data_points_reshaped)
                kde_models[col] = kde
                print(f"KDE model built for {col}")
            except Exception as e:
                print(f"Could not build KDE for {col}: {e}")
        elif len(data_points) == 1:
            print(f"Skipping KDE for {col}: only one data point available, KDE requires at least two. Consider other methods for single-point distributions.")
            kde_models[col] = None # No KDE, will need special handling later
        else:
            print(f"Skipping KDE for {col}: no valid data points available for KDE (all are NaN or infinite).")
            kde_models[col] = None # No KDE model

# Print status of KDE models
print("\nKDE models created for the following columns:")
for col, model in kde_models.items():
    print(f"- {col}: {'Ready' if model is not None else 'Not enough data/skipped'}")

# Display head of the dataframe with new columns to verify again (after parsing fix)
print("\nDataFrame head after improved parsing:")
print(df[
    [col for col in df.columns if 'Lower' in col or 'Upper' in col or col in parsing_configs]
].head())
```

## **Bước 2: Xây dựng hàm mật độ xác suất tổng thể bằng KDE (Kernel Density Estimation)**
Để xác định vùng tập trung thực tế của dữ liệu bên trong các khoảng, ta áp dụng phương pháp ước lượng mật độ hạt nhân không tham số.

**Định nghĩa phương pháp:** KDE làm mượt các khoảng dữ liệu bằng cách phân phối xác suất liên tục thay vì rời rạc hóa thành các điểm đơn lẻ.

**Công thức toán học tổng quát:**$$f_h(x) = \frac{1}{n h} \sum_{i=1}^{n} K\left(\frac{x - x_i}{h}\right)$$
Trong đó:
- $n$: Số lượng quan sát hợp lệ.
- $h$: Độ rộng băng thông (bandwidth) điều chỉnh độ mượt.$K(\cdot)$: Hàm hạt nhân (thường dùng Gaussian Kernel).
- $f(x)$: Hàm mật độ xác suất (PDF) ước lượng được từ toàn bộ tập dữ liệu mẫu.

```python
from scipy.stats import gaussian_kde
import matplotlib.pyplot as plt

# 1. Create smooth simulated distribution data
# Filter out rows with infinite upper bounds for now, as these will be handled in Step 3.
# We also ensure that both lower and upper bounds are not NaN.
bounded_income_df = df[(df['Thu nhập - Upper'] != np.inf) &
                       (df['Thu nhập - Lower'].notna()) &
                       (df['Thu nhập - Upper'].notna())].copy()


# Initialize an empty list to store all generated income points
all_income_points = []

# Number of points to generate for each income range
# This creates a denser representation within each customer's reported range.
num_points_per_range = 100

# Iterate through each row and generate points uniformly within the range
for index, row in bounded_income_df.iterrows():
    lower = row['Thu nhập - Lower']
    upper = row['Thu nhập - Upper']

    # Generate points uniformly within the range
    points = np.random.uniform(lower, upper, num_points_per_range)
    all_income_points.extend(points)

# Convert the list to a NumPy array
all_income_points = np.array(all_income_points)

# 2. Estimate KDE function (f(x))
if len(all_income_points) > 1: # gaussian_kde requires at least 2 data points
    # Use gaussian_kde to estimate the probability density function
    # The 'kde' object can be called like a function to get density values.
    kde = gaussian_kde(all_income_points)
    print("KDE function has been estimated successfully for bounded income ranges.")

    # --- Visualization of the KDE function ---
    # Define a range of values over which to evaluate the KDE
    # Use a slightly wider range than the actual data to show tails better
    min_income_val = all_income_points.min() * 0.9
    max_income_val = all_income_points.max() * 1.1

    x_eval = np.linspace(min_income_val, max_income_val, 1000) # 1000 evaluation points
    density_values = kde(x_eval)

    plt.figure(figsize=(12, 7))
    plt.plot(x_eval, density_values, label='Estimated KDE (f(x))', color='blue')
    plt.fill_between(x_eval, density_values, color='blue', alpha=0.2)
    plt.title('Estimated Probability Density Function (KDE) of Customer Income')
    plt.xlabel('Income (VNĐ)')
    plt.ylabel('Density')
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.legend()
    plt.show()

    # Store the KDE object for later use if needed (e.g., to query f(x))
    # You can now call kde(value) to get the density at 'value'
    # For example, kde(30_000_000) will give the density at 30 million VND.
    df_kde_function = kde # Assign to a variable to be accessible later if needed

else:
    print("Not enough data points to estimate KDE after filtering for bounded ranges.")
```

## **Bước 3: Xử lý phần đuôi mở bằng Phân phối Pareto & Khử giá trị Vô cực (Capping)**
Các khoảng mở trên (như "Trên 100 triệu" hoặc "Trên 5 tỷ") không có cận trên xác định, do đó cần mô hình hóa bằng luật lũy thừa (Power Law).

- **Ước lượng tham số Pareto (MLE):** Xác định ngưỡng bắt đầu đuôi $x_m$ (ví dụ: mốc 100 triệu VNĐ).Tính hệ số hình dáng $\alpha$ bằng phương pháp Hợp lý tối đa:

$$\alpha = n_{tail} \left[ \sum_{i=1}^{n_{tail}} \ln\left(\frac{x_i}{x_m}\right) \right]^{-1}$$

Hàm mật độ phần đuôi:

$$f_{pareto}(x) = \frac{\alpha x_m^\alpha}{x^{\alpha+1}} \quad (\text{với } x \ge x_m)$$

- **Khử giá trị Vô cực (inf) bằng Capping:**
Nếu tham số $\alpha \le 1$, kỳ vọng toán học lý thuyết tiến tới vô cực ($\infty$). Để phục vụ cho báo cáo và trực quan hóa dữ liệu trên Power BI/Excel không bị lỗi vỡ trục, áp dụng giá trị trần an toàn (Capping Value):$$\text{Max Cap} = L_i \times 1.5 \text{ đến } 2.0$$

(Ví dụ: Nhóm trên 100 triệu VNĐ được chặn trần hợp lý ở mức 150.000.000 VNĐ).

```python
from scipy.optimize import minimize_scalar

# --- Step 3: Handling open-ended tails with Pareto distribution and Capping ---

pareto_models = {}
capping_values = {}

print("\n--- Step 3: Handling open-ended tails with Pareto distribution and Capping ---")

for col in columns_to_estimate:
    lower_col = f'{col} - Lower'
    upper_col = f'{col} - Upper'

    if lower_col in df.columns and upper_col in df.columns:
        # Filter for data points with infinite upper bounds (open-ended tails)
        open_ended_mask = (df[upper_col] == np.inf) & df[lower_col].notna()
        tail_data = df[open_ended_mask][lower_col].dropna()

        if len(tail_data) > 0:
            # The minimum value in the tail (x_m) is the threshold for the Pareto distribution
            x_m = tail_data.min()

            if x_m == 0: # Pareto distribution requires x_m > 0
                print(f"Warning: {col} has 0 as a lower bound for an open-ended interval. Pareto model requires x_m > 0. Skipping Pareto for this column.")
                pareto_models[col] = None
                capping_values[col] = None
                continue

            # Likelihood function for Pareto's alpha (MLE for alpha)
            def neg_log_likelihood(alpha, data, x_m_val):
                if alpha <= 0: # Alpha must be positive
                    return np.inf
                # For data >= x_m, the formula is n * log(alpha) + n * alpha * log(x_m) - (alpha + 1) * sum(log(data))
                # Here we simplify sum(log(data)) to directly use the data points
                return -np.sum(np.log(alpha) + alpha * np.log(x_m_val) - (alpha + 1) * np.log(data))

            # Minimize the negative log-likelihood to find alpha
            try:
                # Initial guess for alpha (often around 1 to 2 for real-world data)
                initial_alpha_guess = 1.5
                # Use bounded minimization for alpha (must be > 0)
                result = minimize_scalar(neg_log_likelihood, bounds=(0.01, 10.0), args=(tail_data.values, x_m), method='bounded')
                alpha = result.x

                pareto_models[col] = {'x_m': x_m, 'alpha': alpha}
                print(f"Pareto model built for {col} with x_m={x_m:.2f} and alpha={alpha:.2f}")

                # --- Capping for infinite values ---
                # If alpha <= 1, theoretical expectation is infinite. Apply a capping factor.
                # Using a reasonable heuristic: Lower bound of the 'np.inf' range * capping_factor

                capping_factor = 1.5 if alpha <= 1.0 else 1.2 # Stronger cap for problematic alphas

                # Ensure capping is based on the actual lower bound of the open-ended range
                # For each specific row with np.inf, its cap is based on its *own* lower bound
                # This means we don't apply one global cap, but calculate it per row when needed.

                # Store a function or heuristic to calculate capping value later
                capping_values[col] = {'factor': capping_factor}
                print(f"Capping rule set for {col} (factor: {capping_factor})")

            except Exception as e:
                print(f"Could not fit Pareto model or determine capping for {col}: {e}")
                pareto_models[col] = None
                capping_values[col] = None
        else:
            print(f"No open-ended values (np.inf) found in {col}. Skipping Pareto modeling and capping.")
            pareto_models[col] = None
            capping_values[col] = None

print("\nPareto models and Capping rules established for the following columns:")
for col in columns_to_estimate:
    pareto_status = 'Ready' if pareto_models.get(col) else 'N/A'
    capping_status = 'Ready' if capping_values.get(col) else 'N/A'
    print(f"- {col}: Pareto={pareto_status}, Capping={capping_status}")
```

## **Bước 4: Tính giá trị kỳ vọng có điều kiện (Conditional Expectation) cho từng khách hàng**

Đây là bước cốt lõi tạo ra một con số định lượng duy nhất, độc lập và chính xác cho riêng từng khách hàng dựa trên vị trí của họ trong không gian phân phối.Công thức kỳ vọng có điều kiện giới hạn trong khoảng $[L_i, U_i]$:

$$E_i = E[X \mid L_i \le X \le U_i] = \frac{\int_{L_i}^{U_i} x \cdot f(x) \, dx}{\int_{L_i}^{U_i} f(x) \, dx}$$

**Ý nghĩa các đại lượng:**
- **Tử số ($\int_{L_i}^{U_i} x \cdot f(x) \, dx$)**: Tích phân trọng số giá trị tài chính nhân với mật độ xác suất xuất hiện của nó trong khoảng chặn của khách hàng.

- **Mẫu số ($\int_{L_i}^{U_i} f(x) \, dx$)**: Tích phân hàm mật độ trong giới hạn khoảng để chuẩn hóa tổng diện tích xác suất nội bộ bằng 1.

- **$E_i$**: Giá trị số thực tế ước lượng cuối cùng được gán riêng cho khách hàng $i$ (lưu vào cột dữ liệu mới thay thế cho chuỗi văn bản ban đầu).

```python
from scipy.integrate import quad

def calculate_conditional_expectation(row, col_name, kde_models, pareto_models, capping_values, min_val_for_integration=1e-9):
    lower = row[f'{col_name} - Lower']
    upper = row[f'{col_name} - Upper']

    # Case 1: NaN values for both bounds (e.g., 'Chưa theo dõi cụ thể')
    if pd.isna(lower) and pd.isna(upper):
        return np.nan

    # NEW: If only one bound is NaN, return NaN as we can't integrate a non-finite range
    if pd.isna(lower) or pd.isna(upper):
        return np.nan

    # Case 2: Exact value (lower == upper)
    if lower == upper:
        return lower

    # Determine the effective upper bound, handling np.inf with capping
    effective_upper_for_integration = upper
    if upper == np.inf:
        pareto_config = pareto_models.get(col_name)

        if pareto_config:
            # The lower bound for the current open-ended interval from the data
            current_lower_bound_for_capping = lower

            capping_factor_config = capping_values.get(col_name)
            factor = capping_factor_config['factor'] if capping_factor_config else 1.5 # Default if not found

            # Apply capping based on the current lower bound of the open interval
            effective_upper_for_integration = current_lower_bound_for_capping * factor

            # Ensure the capped value is reasonable and not less than the lower bound
            if effective_upper_for_integration < lower:
                effective_upper_for_integration = lower * 1.5 # Ensure some margin if cap is too low
        else:
            # Fallback for np.inf if no Pareto model was built (e.g., no data for it)
            effective_upper_for_integration = lower * 1.5

    # Ensure lower is not greater than effective_upper_for_integration after all adjustments
    if lower > effective_upper_for_integration:
        # This should ideally not happen with correct parsing and capping, but as a safeguard:
        return lower

    # Get KDE model for the column
    kde = kde_models.get(col_name)

    if kde is None:
        # Fallback to midpoint if no KDE model is available for this column
        return (lower + effective_upper_for_integration) / 2

    # Define the PDF for integration using the KDE model
    def pdf_for_kde(x):
        # Ensure x is an array for kde.evaluate and ensure non-negative output
        return max(0, kde.evaluate(np.array([x]))[0])

    # Integrand for numerator: x * f(x)
    def numerator_integrand(x):
        return x * pdf_for_kde(x)

    # Perform numerical integration
    try:
        # Integrate numerator and denominator over the effective finite interval
        integral_num, _ = quad(numerator_integrand, lower, effective_upper_for_integration)
        integral_den, _ = quad(pdf_for_kde, lower, effective_upper_for_integration)

        if integral_den > min_val_for_integration: # Avoid division by near-zero values
            return integral_num / integral_den
        else:
            # If denominator is too small (very low probability density in interval),
            # fall back to midpoint as a safe heuristic.
            return (lower + effective_upper_for_integration) / 2
    except Exception as e:
        # Handle potential integration errors (e.g., singularities, bad bounds)
        print(f"Integration error for {col_name} between {lower} and {effective_upper_for_integration}: {e}. Falling back to midpoint.")
        return (lower + effective_upper_for_integration) / 2

# Apply the conditional expectation function to create new columns
print("\n--- Step 4: Calculating Conditional Expectation ---")
for col in columns_to_estimate:
    # Create a new column named '[Original Column] - Expectation'
    df[f'{col} - Expectation'] = df.apply(
        lambda row: calculate_conditional_expectation(row, col, kde_models, pareto_models, capping_values),
        axis=1
    )
    print(f"Calculated conditional expectation for {col}")

print("\nConditional expectation calculation complete. Displaying head with new expectation columns:")
# Display relevant columns: original, lower, upper, and expectation
display_cols = []
for col in columns_to_estimate:
    if col in df.columns: display_cols.append(col)
    if f'{col} - Lower' in df.columns: display_cols.append(f'{col} - Lower')
    if f'{col} - Upper' in df.columns: display_cols.append(f'{col} - Upper')
    if f'{col} - Expectation' in df.columns: display_cols.append(f'{col} - Expectation')

print(df[display_cols].head())
```

## **Tạo thêm cột tỷ lệ người phụ thuộc**
Thực hiện gán các giá trị tương ứng ở cột `Có người phụ thuộc tài chính`

- Không có: 0.00
- Có 1-2 người: 0.00
- Có 3-4 ngời: 0.01
- Có từ 5 người: 0.015

```python
import numpy as np

# Create the 'Tỷ lệ người phụ thuộc' column based on the mapping rules
conditions = [
    df['Có người phụ thuộc tài chính'] == 'Không có',
    df['Có người phụ thuộc tài chính'] == 'Có 1-2 người',
    df['Có người phụ thuộc tài chính'] == 'Có 3-4 người', # Corrected typo 'ngời' to 'người'
    df['Có người phụ thuộc tài chính'] == 'Có từ 5 người'
]

choices = [0.00, 0.00, 0.01, 0.015]

df['Tỷ lệ người phụ thuộc'] = np.select(conditions, choices, default=np.nan)

print("New column 'Tỷ lệ người phụ thuộc' created successfully.")
print(df[['Có người phụ thuộc tài chính', 'Tỷ lệ người phụ thuộc']].head())
```

# **II. TÍNH DÒNG TIỀN KHẢ DỤNG**


### **Tính các chỉ số bổ sung:**
- `Chi tiêu` = `Thu nhập - Expectation` * `Tỷ lệ chi tiêu - Expectation`

- `Nghĩa vụ nợ` = `Thu nhập - Expectation `  * `Tỷ lệ thanh toán Nợ`

- `Chi phí người phụ thuộc` = `Thu nhập - Expectation`  * `Tỷ lệ người phụ thuộc`

### **Tính giá trị Dòng tiền khả dụng của từng khách hàng**

- `Dòng tiền khả dụng` = `Thu nhập - Expectation ` - `Chi tiêu` - `Nghĩa vụ nợ` - `Chi phí người phụ thuộc`

```python
df['Chi tiêu'] = df['Thu nhập - Expectation'] * df['Tỷ lệ chi tiêu - Expectation']
df['Nghĩa vụ nợ'] = df['Thu nhập - Expectation'] * df['Tỷ lệ thanh toán Nợ - Expectation']
df['Chi phí người phụ thuộc'] = df['Thu nhập - Expectation'] * df['Tỷ lệ người phụ thuộc']

# Calculate 'Dòng tiền khả dụng'
# Treat NaN values in expense components as 0 for the purpose of Dòng tiền khả dụng calculation
df['Dòng tiền khả dụng'] = (
    df['Thu nhập - Expectation'] -
    df['Chi tiêu'].fillna(0) -
    df['Nghĩa vụ nợ'].fillna(0) -
    df['Chi phí người phụ thuộc'].fillna(0)
)

print("Calculated 'Chi tiêu', 'Nghĩa vụ nợ', 'Chi phí người phụ thuộc', and 'Dòng tiền khả dụng'.")
display(df[['Thu nhập - Expectation', 'Tỷ lệ chi tiêu - Expectation', 'Tỷ lệ thanh toán Nợ - Expectation', 'Tỷ lệ người phụ thuộc', 'Chi tiêu', 'Nghĩa vụ nợ', 'Chi phí người phụ thuộc', 'Dòng tiền khả dụng']].head())
```

## **Đánh giá tương quan dòng tiền**
Tạo cột "Đánh giá dòng tiền": So sánh giá trị Dòng tiền khả tiền khả dụng với khoảng giá trị của cột Tích luỹ.
- Nếu giá trị Dòng tiền khả dụng lớn hơn hoặc nằm trong khoảng tích luỹ thì gán giá trị "Hợp lý"
- Nếu Nếu giá trị Dòng tiền khả dụng nhỏ hơn giá trị Tích luỹ thì gán giá trị "Không hợp lý"
- Nếu trường hợp giá trị cột tích luỹ là NaN mà giá trị của Dòng tiền là các số âm gần -1 hoặc -2 thì vẫn gán giá trị "Hợp lý"


```python
import numpy as np

# Define conditions and choices for np.select
conditions = []
choices = []

# Pre-calculate common boolean series for readability
savings_lower_is_nan = df['Tích luỹ - Lower'].isna()
savings_is_zero = (df['Tích luỹ - Lower'] == 0) & (df['Tích luỹ - Upper'] == 0)
cash_flow_is_negative = (df['Dòng tiền khả dụng'] < 0)
# CAP_NEGATIVE_CASH_FLOW is assumed to be defined in a previous cell.
CAP_NEGATIVE_CASH_FLOW = -2_000_000  # Define the missing variable here
cash_flow_near_neg_bounds = (df['Dòng tiền khả dụng'] >= CAP_NEGATIVE_CASH_FLOW) & cash_flow_is_negative

# Condition 1 (Revised priority): If 'Tích luỹ' is 0, then 'Hợp lý' (covers both positive and negative cash flow)
conditions.append(savings_is_zero)
choices.append("Hợp lý")

# Condition 2 (from original): Tích luỹ is NaN AND Dòng tiền khả dụng is negative but within [-2,000,000, 0) -> "Hợp lý"
conditions.append(savings_lower_is_nan & cash_flow_near_neg_bounds)
choices.append("Hợp lý")

# Condition 3 (from original): Tích luỹ is NOT NaN and NOT ZERO AND Dòng tiền khả dụng is greater than or equal to Tích luỹ - Lower -> "Hợp lý"
conditions.append(~savings_lower_is_nan & ~savings_is_zero & (df['Dòng tiền khả dụng'] >= df['Tích luỹ - Lower']))
choices.append("Hợp lý")

# Condition 4 (from original): Tích luỹ is NOT NaN and NOT ZERO AND Dòng tiền khả dụng is less than Tích luỹ - Lower -> "Không hợp lý"
conditions.append(~savings_lower_is_nan & ~savings_is_zero & (df['Dòng tiền khả dụng'] < df['Tích luỹ - Lower']))
choices.append("Không hợp lý")

# Apply np.select to create the new column 'Đánh giá dòng tiền'
# Default to None if no condition met, allowing for object dtype which can hold strings and NaNs
df['Đánh giá dòng tiền'] = np.select(conditions, choices, default=None)

print("Created new column 'Đánh giá dòng tiền'.")
# Display relevant columns to verify the new assessment, including savings bounds for clarity
display(df[['Dòng tiền khả dụng', 'Tích luỹ', 'Tích luỹ - Lower', 'Tích luỹ - Upper', 'Đánh giá dòng tiền']].head(10))
```

# **III. Tính số vốn khả dụng hiện có**
Tạo cột **Vốn khả dụng** và tính giá trị của cột đó teo công thức:
`Vốn khả dụng = Tài sản nhàn rỗi - Expectation + Dòng tiền khả dụng`

- Nếu Dòng tiền khả dụng có giá trị bé hơn 0 thì `Vốn khả dụng = Tài sản nhàn rỗi - Expectation`
- Ngược lại, nếu Tài sản nhàn rỗi - Expectation có giá trị trống thì `Vốn khả dụng = Dòng tiền khả dụng`

```python
import numpy as np

# Calculate 'Vốn khả dụng' based on the conditions

# Define boolean series for readability
is_taisan_nan = df['Tài sản nhàn rỗi - Expectation'].isna()
is_dongtien_negative = df['Dòng tiền khả dụng'] < 0

conditions = [
    is_taisan_nan, # Condition 1: If 'Tài sản nhàn rỗi - Expectation' is NaN
    ~is_taisan_nan & is_dongtien_negative, # Condition 2: If 'Tài sản nhàn rỗi - Expectation' is NOT NaN AND 'Dòng tiền khả dụng' is negative
    ~is_taisan_nan & ~is_dongtien_negative # Condition 3: If 'Tài sản nhàn rỗi - Expectation' is NOT NaN AND 'Dòng tiền khả dụng' is non-negative
]

choices = [
    df['Dòng tiền khả dụng'], # Choice for Condition 1
    df['Tài sản nhàn rỗi - Expectation'], # Choice for Condition 2
    df['Tài sản nhàn rỗi - Expectation'] + df['Dòng tiền khả dụng'] # Choice for Condition 3
]

df['Vốn khả dụng'] = np.select(conditions, choices, default=np.nan)

print("Đã tạo cột 'Vốn khả dụng'.")
# Display relevant columns to verify the new assessment
display(df[['Tài sản nhàn rỗi - Expectation', 'Dòng tiền khả dụng', 'Vốn khả dụng']].head())
```

## **Đánh giá tương quan Số tiền sẵn sàng đầu tư**
Tạo cột "Đánh giá đầu tư": So sánh giá trị Vốn khả dụng với khoảng giá trị của cột Số tiền ban đầu:
- Nếu giá trị Vốn khả dụng lớn hơn hoặc nằm trong khoảng Số tiền ban đầu thì gán giá trị **"Tốt"**
- Nếu Nếu giá trị Vốn tiền khả dụng nhỏ hơn giá trị Số tiền ban đầu thì gán giá trị **"Không tốt"**

```python
import numpy as np

# Define conditions and choices for np.select for 'Đánh giá đầu tư'
conditions_investment = [
    df['Vốn khả dụng'] >= df['Số tiền ban đầu - Lower'],
    df['Vốn khả dụng'] < df['Số tiền ban đầu - Lower']
]

choices_investment = [
    "Tốt",
    "Không tốt"
]

# Apply np.select to create the new column 'Đánh giá đầu tư'
df['Đánh giá đầu tư'] = np.select(conditions_investment, choices_investment, default=None)

print("Đã tạo cột 'Đánh giá đầu tư'.")
# Display relevant columns to verify the new assessment
display(df[['Vốn khả dụng', 'Số tiền ban đầu - Lower', 'Số tiền ban đầu - Upper', 'Đánh giá đầu tư']].head())
```
